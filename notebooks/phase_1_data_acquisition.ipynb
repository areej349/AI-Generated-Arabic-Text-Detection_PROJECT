```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("KFUPM-JRCAI/arabic-generated-abstracts")

df = pd.DataFrame(dataset["by_polishing"])
df.head()
```

```python
print("\n info col")
print(df.dtypes)
print("\n describe data")
print(df.info())
```

```python

text_col = df.columns[0]   

df["text_length"] = df[text_col].astype(str).str.len()

df["text_length"].describe()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_length_comparison(df, split_name):
    lengths = {
        "Human": df["original_abstract"].astype(str).str.len().mean(),
        "Allam (AI)": df["allam_generated_abstract"].astype(str).str.len().mean(),
        "Jais (AI)": df["jais_generated_abstract"].astype(str).str.len().mean(),
        "LLaMA (AI)": df["llama_generated_abstract"].astype(str).str.len().mean(),
        "OpenAI (AI)": df["openai_generated_abstract"].astype(str).str.len().mean(),
    }

    plt.figure(figsize=(8, 5))
    plt.bar(lengths.keys(), lengths.values())
    plt.title(f"Average Text Length Comparison ({split_name})")
    plt.ylabel("Average Number of Characters")
    plt.xticks(rotation=30)


    plt.savefig(
        rf"C:\Users\areej\FinalProject(AI_Generated_Arabic_Text_Detection)\reports\figures\text_length_{split_name.lower().replace(' ', '_')}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# استدعاء الدالة لكل split
plot_length_comparison(pd.DataFrame(dataset["by_polishing"]), "By Polishing")
plot_length_comparison(pd.DataFrame(dataset["from_title"]), "From Title")
plot_length_comparison(pd.DataFrame(dataset["from_title_and_content"]), "From Title and Content")
```
